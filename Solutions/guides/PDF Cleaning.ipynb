```python
import os
try:
    from reportlab.lib.pagesizes import letter
    from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Table, TableStyle, PageBreak
    from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
    from reportlab.lib import colors
except ImportError:
    print("Please install reportlab first using: pip install reportlab")
    exit()

def create_pdf():
    pdf_filename = "Final_Report_of_ERT_Revised.pdf"
    doc = SimpleDocTemplate(pdf_filename, pagesize=letter,
                            rightMargin=40, leftMargin=40, topMargin=40, bottomMargin=40)
    story = []
    styles = getSampleStyleSheet()
    
    # Custom Typography Styles
    title_style = ParagraphStyle('CoverTitle', parent=styles['Heading1'], fontName='Helvetica-Bold', fontSize=22, leading=26, alignment=1, spaceAfter=15)
    subtitle_style = ParagraphStyle('CoverSubtitle', parent=styles['Normal'], fontName='Helvetica', fontSize=12, leading=16, alignment=1, spaceAfter=30)
    h1_style = ParagraphStyle('SectionH1', parent=styles['Heading2'], fontName='Helvetica-Bold', fontSize=14, leading=18, spaceBefore=15, spaceAfter=10, keepWithNext=True)
    body_style = ParagraphStyle('BodyTextCustom', parent=styles['BodyText'], fontName='Helvetica', fontSize=10, leading=14, spaceAfter=8)
    bullet_style = ParagraphStyle('BulletCustom', parent=styles['Bullet'], fontName='Helvetica', fontSize=10, leading=14, spaceAfter=4, bulletIndent=10, leftIndent=25)
    
    # --- COVER PAGE ---
    story.append(Spacer(1, 40))
    story.append(Paragraph("REPORT ON", subtitle_style))
    story.append(Paragraph("ELECTRICAL RESISTIVITY TEST FOR PROPOSED STRUCTURE", title_style))
    story.append(Paragraph("AT", subtitle_style))
    story.append(Paragraph("<b>client, Location SITE</b>", title_style))
    story.append(Spacer(1, 40))
    story.append(Paragraph("CONDUCTED BY", subtitle_style))
    story.append(Paragraph("<b>BOSE ENGINEERS</b><br/>53, PURNA CHANDRA MITRA LANE<br/>KOLKATA – 700033", subtitle_style))
    story.append(Paragraph("<i>A NABL ACCREDITED LABORATORY<br/>An ISO 9001-2015 Certified Organization</i>", subtitle_style))
    story.append(Spacer(1, 40))
    
    job_data = [
        [Paragraph("<b>Job No.</b>", body_style), Paragraph("<b>Date</b>", body_style), Paragraph("<b>Prepared By</b>", body_style), Paragraph("<b>Checked By</b>", body_style), Paragraph("<b>Approved By</b>", body_style)],
        ["BE-2024-231", "March 2025", "R. Chowdhury", "B. Bose", "Dr. S.K. Bose"]
    ]
    t_job = Table(job_data, colWidths=[90, 80, 110, 110, 110])
    t_job.setStyle(TableStyle([
        ('BACKGROUND', (0,0), (-1,0), colors.lightgrey),
        ('ALIGN', (0,0), (-1,-1), 'CENTER'),
        ('GRID', (0,0), (-1,-1), 0.5, colors.grey),
        ('PADDING', (0,0), (-1,-1), 6),
    ]))
    story.append(t_job)
    story.append(PageBreak())
    
    # --- PAGE 2: MAIN REPORT CONTENT ---
    story.append(Paragraph("<b>Project:</b> Electrical Resistivity Test for Proposed Outsourced Semiconductor Assembly and Test (OSAT) Facility at Jagi Road, Assam.", body_style))
    story.append(Spacer(1, 10))
    
    story.append(Paragraph("1. INTRODUCTION", h1_style))
    story.append(Paragraph("Electrical Resistivity Test (ERT) was carried out at 2 (Two) numbers prefixed location for at Proposed Structure at <b>client, Location</b>.", body_style))
    story.append(Paragraph("The tests were conducted in January 2025.", body_style))
    
    story.append(Paragraph("2. METHODOLOGY", h1_style))
    story.append(Paragraph("During this test Wenner Method electrodes configuration as per IS: 3043-1987 in 8 directions was employed, with gradual spacing between electrodes. All four electrodes were placed in one line; the outer two electrodes were used for current electrodes through which current is inserted into the ground, whereas the inner two electrodes were used as potential electrodes for measuring potential difference. Distance between each two consecutive electrodes (S) was kept equal and the depth of burial of electrodes was 10cm to 15cm.", body_style))
    
    story.append(Paragraph("As per clause 36.3.1.1 of IS 3043-1987 depth of burial of electrodes in the ground 'd' is negligible compared to spacing between the electrodes, then:", body_style))
    story.append(Paragraph("<b>Resistivity = 2 x π x S x V/I Ohm-m</b>", ParagraphStyle('Formula', parent=body_style, alignment=1, fontName='Helvetica-Bold')))
    
    story.append(Paragraph("• <b>S</b> = Distance between two successive electrodes in meters", bullet_style))
    story.append(Paragraph("• <b>V</b> = Voltage difference between two inner electrodes in Volt", bullet_style))
    story.append(Paragraph("• <b>I</b> = Current flowing through the two outer electrodes in ampere.", bullet_style))
    
    story.append(Paragraph("As per clause 36.3.1.2 of IS 3043-1987, resistivity is given by <b>ρ = 2.π.S.R</b> where ρ = Average Apparent Resistivity in ohm-cm, π = 3.1415927, S = Distance in meter between two consecutive electrodes and corresponds to a point at the center of electrodes spreads and R = Resistance in ohms.", body_style))
    
    story.append(Spacer(1, 15))
    story.append(Paragraph("3. LOCATION OF TEST", h1_style))
    story.append(Paragraph("The locations of test are tabulated in Table 1 as below:", body_style))
    
    t1_data = [
        [Paragraph("<b>ERT No.</b>", body_style), Paragraph("<b>Location</b>", body_style)],
        ["ERT-1", "Boiler"],
        ["ERT-2", "TG Deck"]
    ]
    t1 = Table(t1_data, colWidths=[150, 200])
    t1.setStyle(TableStyle([('BACKGROUND', (0,0), (-1,0), colors.lightgrey), ('GRID', (0,0), (-1,-1), 0.5, colors.grey), ('PADDING', (0,0), (-1,-1), 6)]))
    story.append(t1)
    
    story.append(Spacer(1, 15))
    story.append(Paragraph("4. RESULTS", h1_style))
    story.append(Paragraph("Summarized results of Earth Resistivity Test (ERT) are given in the Table below.", body_style))
    
    t2_data = [
        [Paragraph("<b>ERT No.</b>", body_style), Paragraph("<b>Mean Resistance (Ohm)</b>", body_style), Paragraph("<b>Mean Resistivity (Ohm-M)</b>", body_style)],
        ["ERT-1", "0.796", "29.37"],
        ["ERT-2", "2.206", "57.04"]
    ]
    t2 = Table(t2_data, colWidths=[120, 150, 150])
    t2.setStyle(TableStyle([('BACKGROUND', (0,0), (-1,0), colors.lightgrey), ('GRID', (0,0), (-1,-1), 0.5, colors.grey), ('PADDING', (0,0), (-1,-1), 6)]))
    story.append(t2)
    
    story.append(Spacer(1, 15))
    story.append(Paragraph("<i>Project Chart Details: GEOTECHNICAL INVESTIGATION FOR Proposed Structure at <b>client, Location</b>, West Bengal, India. (POLAR CURVE # ERT-1 & ERT-2)</i>", body_style))
    
    doc.build(story)
    print(f"Success! '{pdf_filename}' generated in your directory.")

if __name__ == "__main__":
    create_pdf()
```
