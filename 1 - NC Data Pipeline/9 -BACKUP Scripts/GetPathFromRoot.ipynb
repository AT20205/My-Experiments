```python
import os
from pathlib import Path


def list_paths_from_env(env_var_name: str) -> list[str]:
    """Retrieves a root directory from an environment variable and recursively lists all child paths."""
    # Read the environment variable
    env_path = os.getenv(env_var_name)

    if not env_path:
        print(f"Error: Environment variable '{env_var_name}' is not set.")
        return []

    # Convert to a Path object (pathlib handles slash orientation automatically)
    base_path = Path(env_path)

    if not base_path.exists():
        print(
            f"Error: The directory '{base_path}' specified in '{env_var_name}' does not exist."
        )
        return []

    # Recursively fetch all files and folders
    return [str(p) for p in base_path.rglob("*")]


ENV_VAR_NAME = "root"

paths = list_paths_from_env(ENV_VAR_NAME)

for p in paths:
    print(p)
```
